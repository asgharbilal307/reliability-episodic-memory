## Purpose

This notebook performs complete EDA on the task dataset and synthetic memory bank. Every visualisation that appears in the Phase 1 report is generated here with accompanying interpretation. The key questions we answer:

1. Is the task dataset balanced across bug categories?
2. Do complexity and function length differ meaningfully across categories?
3. What is the reliability distribution of the synthetic memory bank?
4. Does the confidence estimator look plausibly calibrated on synthetic data?
5. How does the recency decay function behave over time?

```python
import json
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import math

PALETTE = {'off_by_one':'#4C72B0','type_error':'#DD8452','scope_error':'#55A868','logic_error':'#C44E52'}
CATS = ['off_by_one','type_error','scope_error','logic_error']
CAT_LABELS = ['Off-by-One','Type Error','Scope Error','Logic Error']

with open('../data/humaneval_subset.json') as f:
    problems = json.load(f)
with open('../data/synthetic_memories.json') as f:
    memories = json.load(f)

df = pd.DataFrame(problems)
df_mem = pd.DataFrame(memories).drop(columns=['embedding_placeholder'])

print(f'Problems loaded: {len(df)}')
print(f'Memories loaded: {len(df_mem)}')
```

## 1. Task Dataset: Class Balance

**Why this matters:** The three-agent comparison in Phase 2 must be fair. If one bug category dominates, per-category results will be noisy for the minority class. We confirm the dataset is balanced before running any agent.

```python
from IPython.display import Image
Image('../images/fig1_bug_distribution.png')
```

```python
print('Category counts:')
print(df['bug_category'].value_counts())
print(f'\nBalance check -- all equal to 20: {all(v==20 for v in df["bug_category"].value_counts())}')
```

**Finding:** The dataset is **perfectly balanced** (20 problems per category). No class imbalance correction is needed. This means per-category success rates in Phase 2 are directly comparable without reweighting.

## 2. Function Length Distribution by Category

```python
Image('../images/fig2_function_lengths.png')
```

```python
print('Mean function length by category:')
print(df.groupby('bug_category')['function_length'].mean().round(2))
print('\nMax function length by category:')
print(df.groupby('bug_category')['function_length'].max())
```

**Finding:** Most problems are short (2-6 lines), keeping per-episode token cost manageable for 100-episode runs on Groq (Llama 3.1 8B). Logic errors contain the longest functions (up to 16 lines), which is consistent with logic bugs often arising in multi-branch code. This length difference will be controlled for in Phase 2 per-category analysis.

## 3. Cyclomatic Complexity Distribution

```python
Image('../images/fig3_complexity_boxplot.png')
```

```python
print('Complexity stats by category:')
print(df.groupby('bug_category')['complexity'].describe().round(2))
```

**Finding:** Scope errors and logic errors have higher median complexity than off-by-one and type errors. This is expected: scope bugs typically arise in nested function definitions or closures, and logic errors often involve multi-branch conditionals. The reliability-aware agent should show greater benefit on high-complexity problems, since memories from similar high-complexity past tasks are more likely to be reliably reusable.

## 4. Function Length vs Complexity Correlation

```python
Image('../images/fig4_length_vs_complexity.png')
```

```python
from scipy import stats
all_x = df['function_length'].values
all_y = df['complexity'].values
slope, intercept, r, p, se = stats.linregress(all_x, all_y)
print(f'Pearson r = {r:.3f}')
print(f'R² = {r**2:.3f}')
print(f'p-value = {p:.4f}')
```

**Finding:** There is a **moderate positive correlation** (r ≈ 0.6) between function length and cyclomatic complexity. This is unsurprising — longer functions tend to have more branches. The relationship is not strong enough to treat length as a proxy for complexity; both features are retained in the dataset schema. For the agent, this means that longer functions will tend to be harder, but complexity remains an independent predictor of difficulty.

## 5. Synthetic Memory Bank: Reliability Distribution

```python
Image('../images/fig5_memory_sr_dist.png')
```

```python
srs = df_mem['success_rate'].values
print(f'Mean SR:   {np.mean(srs):.3f}')
print(f'Std SR:    {np.std(srs):.3f}')
print(f'SR < 0.3:  {(srs < 0.3).sum()} memories (low reliability)')
print(f'SR >= 0.7: {(srs >= 0.7).sum()} memories (high reliability)')
```

**Finding:** The memory bank has 20 high-reliability (SR >= 0.7) and 5 low-reliability (SR < 0.3) memories. The 5 low-reliability memories are essential for measuring the **false memory rate** metric: Baseline B (similarity-only) will sometimes retrieve these over better options when semantic similarity is high, while the reliability-aware agent should learn to down-rank them.

## 6. Confidence Calibration Preview

```python
Image('../images/fig6_calibration_preview.png')
```

```python
from scipy.stats import spearmanr
conf = df_mem['confidence'].values
sr   = df_mem['success_rate'].values
rho, p = spearmanr(conf, sr)
print(f'Spearman correlation (confidence vs SR): rho={rho:.3f}, p={p:.4f}')
```

**Finding:** On synthetic data, confidence and success rate are highly correlated (Spearman rho > 0.95), because we set confidence proportional to SR when constructing the bank. This is a **design check**, not an empirical result. The real calibration question — whether the LLM's self-estimated confidence after real episodes matches empirical SR — is the ECE measurement in Phase 2 (Weeks 7-8). The near-perfect calibration here just confirms the synthetic bank behaves as designed.

## 7. Recency Decay Behaviour

```python
Image('../images/fig7_recency_decay.png')
```

```python
lam = 0.01
for days in [0, 7, 30, 60, 90, 180]:
    r = math.exp(-lam * days)
    print(f'{days:>4}d  R(m) = {r:.4f}')
```

**Finding:** With λ=0.01, a memory from 30 days ago retains ~74% of its recency weight; at 90 days it retains ~41%. This decay is slow enough that memories do not expire prematurely during a 100-episode run (which will span days to a week in real time). λ is an ablation hyperparameter — the factorial ablation in Phase 2 will evaluate whether recency decay improves performance over the Sim+SR variant.

## 8. Memory SR and Confidence by Category

```python
Image('../images/fig8_sr_confidence_heatmap.png')
```

```python
print('Mean SR and confidence by category (synthetic memories):')
print(df_mem.groupby('task_category')[['success_rate','confidence']].mean().round(3))
```

**Finding:** All four categories have similar mean SR and confidence in the synthetic bank (by design). This avoids confounding the category variable with reliability. In Phase 2, the actual SR values will emerge from real agent-episode outcomes and will almost certainly vary across categories — that variation is the substance of the per-category analysis in Section 5 of the paper.

## 9. EDA Summary and Implications for Modelling

| Finding | Modelling implication |
|---|---|
| Perfectly balanced categories | Per-category success rates in Phase 2 are directly comparable |
| Short functions (mean 4.7 lines) | Low token cost per episode; 100-episode runs are feasible on Groq free tier |
| Complexity varies by category | Include complexity as a covariate in per-category regression |
| 20 high-SR / 5 low-SR memories | False memory rate metric is measurable in Phase 2 |
| SR-confidence correlation near 1 on synthetic data | Calibration design is correct; real ECE will be the interesting Phase 2 result |
| Recency decay retains 74% at 30 days | λ=0.01 is appropriate for the Phase 2 run duration |
